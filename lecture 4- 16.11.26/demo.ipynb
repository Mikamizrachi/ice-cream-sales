```python
import os
with open("data2.csv", "w") as f:
    f.write('x, y \n ')
    for val in range(101):
        f.write(f'{val}, {val**2}\n')

os.startfile('data2.csv')
```

```python
x_values = []
y_values = []

with open('data2.csv', 'r') as f:
    next(f)
    for line in f:
        parts = line.split(',')
        x_values.append(int(parts[0]))
        y_values.append(int(parts[1]))
        
```

```python
x_values
```

```python
y_values
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(x_values,y_values, linewidth = 10, color = 'red')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.show()
```
